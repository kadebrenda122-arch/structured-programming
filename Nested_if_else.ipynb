```python
s1 = int(input("enter mark for subject 1:"))
s2 = int(input("enter mark for subject 2:"))
s3 = int(input("enter mark for subject 3:"))
s4 = int(input("enter mark for subject 4:"))
s5 = int(input("enter mark for subject 5:"))
avg=(s1+s2+s3+s4+s5)/5
print('average of five subject marks is: '+str(avg))
if(avg>=80):
    print("an outstanding student")
elif(avg>=70):
    print("a smart student")
elif(avg>=60):
    print("average student")
elif(avg>=50):
    print("below-average student")
elif(avg>=40):
    print("a poor student")
else:
    print("need an extra ordinary effort")
```
