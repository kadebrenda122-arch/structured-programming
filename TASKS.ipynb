1. create temperatures. txt with daily readings

```python
def load_temperatures(filename):
    temperatures = []
    try:
        with open(filename,"r") as f:
            for line in f:
                line = line.strip()
                if line != "":
                    temperatures.append(float(line))
    except FileNotFoundError:
        print("Dataset not found:",{filename})    
    return temperatures  

temperatures = load_temperatures("temperatures.txt") 
print("n =", len(temperatures))
print("mean =", sum(temperatures)/len(temperatures)) 
print("max =", max(temperatures)) 
print("min =", min(temperatures))           
```

2. calculating pass rate at 50% and distinction at 80

```python
with open("marks.txt", "r") as f:
    marks = []
    for line in f:
        line = line.strip()
        if line != "":
            marks.append(int(line))
total_students = len(marks)
passed = sum(1 for mark in marks if mark >= 50)
distinctions = sum( 1 for mark in marks if mark>= 80)
if total_students > 0:
    pass_rate= (passed/ total_students)* 100
    distinction_rate = (distinctions/total_students)*100
    print("students who passed:", total_students) 
    print("pass_rate:", pass_rate,"%")
    print("students with distinctions:", distinctions)
    print("Distinction_rate:", distinction_rate, "%")   
else:
    print("No marks in the file")

    
```

4. print a report for my marks data

```python
with open("marks.txt", "r") as f:
    marks = []
    for line in f:
        line = line.strip()
        if line != "":
            marks.append(int(line))
total_students = len(marks)
passed = sum(1 for mark in marks if mark >= 50)
distinctions = sum( 1 for mark in marks if mark>= 80)
if total_students > 0:
    pass_rate= (passed/ total_students)* 100
    distinction_rate = (distinctions/total_students)*100
    print("\n======= MARKS REPORT ========")
    print("students who passed:", total_students) 
    print(f"pass_rate:{pass_rate:.2f}%")
    print("students with distinctions:", distinctions)
    print(f"Distinction_rate: {distinction_rate:.2f}%") 
    print("============================================")
else:
    print("No marks found in the file")

    
```

3. Read sales.txt, print total revenue and items with qty >10

```python
# SEQUENCE
attendance = []
system_running = True
print ("welcome to the library Attendance system")
print("Type quit to exit and generate the final report.\n")
# ITERATION
while system_running:
    student_name = input("Enter student name or ID(or type 'quit'):").strip()
# Selection
    if student_name.lower() == 'quit':
        system_running = False
    elif student_name == "":
        print("Error: Name cannot be blank. Please enter name")
    elif student_name in attendance:
        print(f"Notice: {student_name} is already signed in!.\n")

    else:
        attendance.append(student_name)
        print(f"Success: {student_name} has already been signed in")
print("\n" + "="*30)   
print("FINAL ATTENDANCE REPORT")
print("="*30) 
print(f"Total students in the library: {len(attendance)}")
for index, student in enumerate(attendance,start=1):
    print(f" {index}. {student}")
```
