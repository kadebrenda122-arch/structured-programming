```python
import matplotlib as plt
import seaborn as sns
```

```python
sns.scatterplot(data=global_unemployment,x="2014",y="2015")
```

```python
import plotly.express as px
```

```python
fig = px.pie(global_unemployment,names="age_categories",values="2023",title="age distribution")
fig.show()
```

 This piechart shows the distribution of age_categories in 2023 globally revealing a highly youthful population implying high dependency since there is a small work force represented by the 19.8% of the adults, aswellas future growth predicting high demand for education and entry-level jobs due to the rapidly increasing population.

# first melt years into one column

```python
global_unemployment_melt = global_unemployment.melt(id_vars=['country'], value_vars=['2014','2015','2024'],var_name = 'year',value_name='Unemployment')
sns.lineplot(x='year', y= 'Unemployment',data=global_unemployment_melt)
```

```python
top_10 = global_unemployment.sort_values('2024',ascending=False).head(10)
sns.barplot(x='2024', y='country',data=top_10)
```
